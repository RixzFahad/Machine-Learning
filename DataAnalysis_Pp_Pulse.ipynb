```python
#after creating the clone of phonepe h=github then import the libraries

!git clone https://github.com/PhonePe/pulse.git
import pandas as pd
import json
import os #operating system used to access each file directory path

#take path1 as a aggregated transaction data and acces statewise data

path1 = "/content/pulse/data/aggregated/transaction/country/india/state"

agg_state_list = os.listdir(path1)

#agg_state_list

#extract the data to create the dataframe

clm = {'State':[],'Year':[],'quarter':[],'Transaction_type':[],'Transaction_count':[],'Transaction_amount':[]}

for i in agg_state_list:

    #print(i)

    p_i = path1+"/"+i

    agg_yr = os.listdir(p_i)

    #print(agg_yr)

    for j in agg_yr:

        p_j = p_i + "/" + j

        agg_year_quarter = os.listdir(p_j)

        #print(agg_year_quarter)
        for k in agg_year_quarter:

            p_k = p_j + "/" + k

            #print(p_k)

            data = open(p_k,'r')
            D = json.load(data)
            for z in D['data']['transactionData']:

              Name = z['name']
              count = z['paymentInstruments'][0]['count']
              amount = z['paymentInstruments'][0]['amount']
              clm['Transaction_type'].append(Name)
              clm['Transaction_count'].append(count)
              clm['Transaction_amount'].append(amount)
              clm['State'].append(i)
              clm['Year'].append(j)
              clm['quarter'].append(int(k.strip('.json')))

print(clm)
```

```python
agg_transactions = pd.DataFrame(clm)

agg_transactions.head()
```

# ***Aggregated Insaurance***

```python
path2 = "/content/pulse/data/aggregated/insurance/country/india/state"

agg_insurence_state_list = os.listdir(path2)

print(agg_insurence_state_list)

col2 = {
    'State':[],
    'Year':[],
    'Quarter':[],
    'insurence_type':[],
    'insurence_count':[],
    'insurence_amount':[]
}

for i in agg_insurence_state_list:

    p_i = path2 + "/" + i

    agg_yr = os.listdir(p_i)

    for j in agg_yr:

        p_j = p_i + "/" + j

        agg_year_quarter = os.listdir(p_j)

        for k in agg_year_quarter:

            p_k = p_j + "/" + k

            data = open(p_k,'r')

            D = json.load(data)

            for z in D['data']['transactionData']:

                Name = z['name']

                count = z['paymentInstruments'][0]['count']

                amount = z['paymentInstruments'][0]['amount']

                col2['insurence_type'].append(Name)

                col2['insurence_count'].append(count)

                col2['insurence_amount'].append(amount)

                col2['State'].append(i)

                col2['Year'].append(j)

                col2['Quarter'].append(int(k.strip('.json')))

print(col2)
```
