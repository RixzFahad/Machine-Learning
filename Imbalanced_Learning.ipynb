### Oversampling and Undersampling Techniques

#### Why do we need them?
In **Imbalanced Datasets**, one class (Majority Class) has many more samples than the other class (Minority Class).  
Example: Fraud detection → 99.9% Genuine, 0.01% Fraud.

Standard models become biased towards the majority class → Poor performance on minority class.

---

### 1. Oversampling
**Definition:** Increasing the number of samples in the minority class to balance the dataset.

**How it works:**
- Duplicate existing minority class samples
- Or generate synthetic samples

**Popular Techniques:**
- **Random Oversampling** → Randomly duplicate minority class samples
- **SMOTE (Synthetic Minority Over-sampling Technique)** → Creates new synthetic samples by interpolating between existing minority samples
- **ADASYN** → Adaptive Synthetic Sampling (focuses on difficult-to-learn minority samples)

**Advantages:**
- No data loss
- Better performance on minority class

**Disadvantages:**
- Can cause overfitting (especially Random Oversampling)
- Increases training time and memory usage

---

### 2. Undersampling
**Definition:** Reducing the number of samples in the majority class to balance the dataset.

**How it works:**
- Randomly remove samples from majority class

**Popular Techniques:**
- **Random Undersampling** → Randomly delete majority class samples
- **Tomek Links** → Removes samples near the decision boundary
- **NearMiss** → Selects majority samples based on distance to minority samples
- **Cluster Centroids** → Replaces majority class clusters with centroids

**Advantages:**
- Reduces training time
- Helps model learn minority class better

**Disadvantages:**
- Loss of important information from majority class
- Can lead to underfitting

---

### When to Use Which?

| Situation                        | Recommended Technique      |
|----------------------------------|---------------------------|
| Very small dataset               | Oversampling (SMOTE)      |
| Large dataset                    | Undersampling             |
| High risk of overfitting         | Undersampling             |
| Want to keep all data            | Oversampling              |
| Best overall performance         | **Hybrid** (SMOTE + Tomek / SMOTE + Edited Nearest Neighbors) |

---

**Key Tip:**
- Always split data into train-test **before** applying sampling techniques.
- Apply sampling **only on training data**, never on test data.
- Use **Cross-Validation** to evaluate properly.

**Best Practice:**  
Use **SMOTE** for oversampling and combine it with undersampling techniques for best results in most real-world imbalanced problems.

*Random Oversampling (ROS) and Random Undersampling (RUS) are NOT good approaches for ML models*

*Why?*
**They can cause information loss (RUS) or overfitting (ROS)**

*To overcome ROS and RUS, we use smarter methods:*

***1. Smart Undersampling (Cluster-Based Approach)***

*Step 1:*  
**Take the majority class data**

*Step 2:*  
**Apply K-Means Clustering to identify groups (clusters) in the data**

*Step 3:*  
**From each cluster, remove 50% of the data points**

*Result:*  
**Balanced dataset without losing important patterns**

*Why this works?*
***Because we are removing data evenly from all clusters, preserving structure and distribution***

*This method is called:*
***Cluster Centroid Undersampling***

*2. Smart Way of Oversampling*

*q) How can we increase minority class without duplicating the data?*

*a) SMOTE (Synthetic Minority Oversampling Technique)*

*Step 1:*  
**Randomly choose a data point from the minority class**

*Step 2:*  
**Find its K-Nearest Neighbors (KNN)**

*Step 3:*  
**Select one of the nearest neighbors**

*Step 4:*  
**Compute the distance between the two points**

*Step 5:*  
**Create a new synthetic point somewhere between them**

*Result:*  
**New artificial data points are generated instead of duplicating existing ones**

*Why this works?*  
***It reduces overfitting and creates more realistic data distribution***

*Key Idea:*  
***New points are generated along the line between nearest neighbors***

*Conclusion:*  
***Better than random oversampling because it adds variation, not copies***
