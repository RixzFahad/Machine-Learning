# Students are having Maths and Physics Marks

| Student | **Maths** | **Physics** | ***Overall Science Performance (PC1)*** |
|----------|----------:|------------:|----------------------------------------:|
| A | 90 | 92 | 91 |
| B | 80 | 82 | 81 |
| C | 70 | 72 | 71 |

## Observation

**Physics** and **Maths** marks move together.

- Student A scores high in both subjects.
- Student B scores average in both subjects.
- Student C scores lower in both subjects.

Since both columns contain very similar information, storing both may be redundant.

## PCA Intuition

Instead of storing:

- **Maths**
- **Physics**

PCA creates a new feature:

***Overall Science Performance***

| Student | ***Overall Science Performance*** |
|----------|----------------------------------:|
| A | 91 |
| B | 81 |
| C | 71 |

## Why?

**Maths** and **Physics** are highly correlated.

So PCA combines them into a single component that captures most of the information.

### In Simple Words

***PCA = Compressing multiple related features into fewer features while keeping most of the important information.***

### Student Analogy

Instead of saying:

- A scored **90 in Maths** and **92 in Physics**

We simply say:

- A has an ***Overall Science Performance*** of **91**

This reduces the number of columns from **2 → 1** while preserving most of the information.

## How PCA Reduces Information

To reduce the information, PCA checks the ***variance*** of the data.

### What is Variance?

***Variance*** measures how much the data points are spread out from their average value.

- High Variance = More Information
- Low Variance = Less Information

### PCA Rule

PCA keeps the directions with the ***highest variance*** because they contain the most useful information.

PCA removes the directions with ***low variance*** because they contribute very little information.

### Student Marks Example

| Student | Maths | Physics |
|----------|------:|---------:|
| A | 90 | 92 |
| B | 80 | 82 |
| C | 70 | 72 |

Both **Maths** and **Physics** move together.

The largest variation among students is their ***overall science performance***.

Therefore, PCA combines these columns into a new feature:

***Overall Science Performance***

and keeps this feature because it captures the maximum variance in the data.

### Key Point

***PCA reduces dimensions by keeping the components with the highest variance and removing components with low variance.***

# Covariance Matrix

Before PCA can find the principal components, it first calculates the ***Covariance Matrix***.

## What is Covariance?

Covariance tells us how two variables move together.

- Positive Covariance → Both increase or decrease together.
- Negative Covariance → One increases while the other decreases.
- Zero Covariance → No relationship.

## Student Marks Example

| Student | Maths | Physics |
|----------|------:|---------:|
| A | 90 | 92 |
| B | 80 | 82 |
| C | 70 | 72 |

As Maths marks increase, Physics marks also increase.

So the covariance between Maths and Physics is ***positive***.

## Covariance Matrix

A covariance matrix shows the covariance between every pair of features.

|        | Maths | Physics |
|---------|--------|---------|
| Maths | Var(Maths) | Cov(Maths, Physics) |
| Physics | Cov(Physics, Maths) | Var(Physics) |

Or mathematically:

|        | Maths | Physics |
|---------|---------|---------|
| Maths | σ²ₘ | Cov(M,P) |
| Physics | Cov(P,M) | σ²ₚ |

## Why PCA Uses It?

PCA analyzes the covariance matrix to find:

- Which features are strongly related
- Which directions contain the most variance
- Which information can be compressed

## Intuition

Suppose:

- Maths and Physics always move together.

Then PCA says:

> "These columns contain similar information. Let's combine them into one principal component."

The covariance matrix helps PCA discover this relationship.

## PCA Flow

1. Collect data
2. Standardize data (optional but recommended)
3. Calculate ***Covariance Matrix***
4. Find ***Eigenvalues*** and ***Eigenvectors***
5. Select top principal components
6. Transform data into lower dimensions

### Interview Answer

***Covariance Matrix is a square matrix that shows how every feature varies with every other feature. PCA uses it to identify the directions of maximum variance in the dataset.***
